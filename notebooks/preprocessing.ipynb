## Step 4: Standardizing Column Names

In this step, we standardize the column names across all yearly datasets.

Since the World Happiness files use different column names in different years, we need to rename similar columns into one consistent format.

For example:
- `Country` and `Country or region` will become `country`
- `Happiness Score`, `Happiness.Score`, and `Score` will become `happiness_score`
- `Family` and `Social support` will become `social_support`

This step is important before merging the datasets into one final DataFrame.

```python
def standardize_columns(df, year):
    
    df.columns = df.columns.str.strip()
    
    rename_dict = {}

    for col in df.columns:
        col_lower = col.lower()

        if col_lower in ["country", "country or region"]:
            rename_dict[col] = "country"

        elif col_lower in ["happiness score", "happiness.score", "score"]:
            rename_dict[col] = "happiness_score"

        elif col_lower in [
            "economy (gdp per capita)",
            "economy..gdp.per.capita.",
            "gdp per capita"
        ]:
            rename_dict[col] = "gdp_per_capita"

        elif col_lower in ["family", "social support"]:
            rename_dict[col] = "social_support"

        elif col_lower in [
            "health (life expectancy)",
            "health..life.expectancy.",
            "healthy life expectancy"
        ]:
            rename_dict[col] = "life_expectancy"

        
        elif col_lower in [
            "freedom",
            "freedom to make life choices"
        ]:
            rename_dict[col] = "freedom"

        elif col_lower in ["generosity"]:
            rename_dict[col] = "generosity"

        elif col_lower in [
            "trust (government corruption)",
            "trust..government.corruption.",
            "perceptions of corruption"
        ]:
            rename_dict[col] = "corruption"

    df = df.rename(columns=rename_dict)

    df["year"] = year

    return df
```

```python
all_dfs = []

for file in files:
    
    df = pd.read_csv(file)
    
    year = int(file.split("\\")[-1][:4])
    
    df = standardize_columns(df, year)
    
    all_dfs.append(df)

print("Done")
```

```python
merged_df = pd.concat(all_dfs, ignore_index=True)

print(merged_df.shape)
merged_df.head()
```

```python
print(merged_df.columns.tolist())
```

## Step 7: Selecting the Important Columns

After merging the datasets, some extra columns remain because different years contain additional fields such as confidence intervals, ranks, and residual values.

In this step, we keep only the main columns that are relevant for the dashboard and the analysis of happiness factors.

```python
selected_columns = [
    "country",
    "year",
    "happiness_score",
    "gdp_per_capita",
    "social_support",
    "life_expectancy",
    "freedom",
    "generosity",
    "corruption"
]

merged_df = merged_df[selected_columns]

print(merged_df.head())
print(merged_df.columns.tolist())
```

## Step 8: Handling Missing Values

After selecting the important columns, we need to check for missing values.

Missing values can affect analysis and visualization, so we either remove them or fill them using appropriate methods.

In this project, we will:
- Remove rows missing essential values (country, year, happiness_score)
- Fill missing numerical values using the median

```python
print(merged_df.isnull().sum())
```

## Step 9: Cleaning Missing Values

In this step, we clean the dataset by:

- Removing rows where essential fields are missing
- Filling missing numerical values using the median

This ensures the dataset is complete and suitable for analysis.

```python
merged_df = merged_df.dropna(subset=["country", "year", "happiness_score"])

numeric_cols = [
    "happiness_score",
    "gdp_per_capita",
    "social_support",
    "life_expectancy",
    "freedom",
    "generosity",
    "corruption"
]

for col in numeric_cols:
    merged_df[col] = merged_df[col].fillna(merged_df[col].median())

print("Missing values after cleaning:")
print(merged_df.isnull().sum())
```

```python
merged_df = merged_df.drop_duplicates()

merged_df = merged_df.sort_values(by=["year", "country"]).reset_index(drop=True)

print(merged_df.head())
```
